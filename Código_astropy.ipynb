```python
import numpy as np
from astropy.cosmology import FlatLambdaCDM
import astropy.units as u
import matplotlib.pyplot as plt
```

```python
# definindo constantes
c = 2.9*10**5 # velocidade da luz

# redshift
z = np.linspace(0, 2, 1000)
```

```python
cosmo = FlatLambdaCDM(H0=70 * u.km / u.s / u.Mpc, Om0=0.3)

# H(z)
h_lcdm = cosmo.H(z).value

# Dl(z)
dl_lcdm = cosmo.luminosity_distance(z).value

# mu(z)
mu_lcdm = cosmo.distmod(z).value
```

```python
plt.plot(z, h_lcdm, label='LCDM', color='red')
plt.xlabel('z', fontsize=15)
plt.ylabel('H(z)', fontsize=15)
plt.legend(loc='upper left')
plt.title('Parâmetro de Hubble')
#plt.savefig('Hz_Mateus.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

```python
# dl(z)
plt.plot(z, dl_lcdm, label='LCDM', color='red')
plt.xlabel('z', fontsize=15)
plt.ylabel('D_L(z)', fontsize=15)
plt.legend(loc='upper left')
plt.title('Distância de Luminosidade')
#plt.savefig('dLz_Mateus.pdf', format='pdf', bbox_inches='tight')
plt.show()
```

```python
plt.plot(z, mu_lcdm, label='LCDM', color='red')
plt.xlabel('z', fontsize=15)
plt.ylabel('mu(z)', fontsize=15)
plt.legend(loc='upper left')
plt.title('Módulo da distância')
#plt.savefig('muz_Mateus.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
